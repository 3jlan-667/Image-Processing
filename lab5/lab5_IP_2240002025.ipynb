```python
import cv2
import numpy as np
import matplotlib.pyplot as plt
# Read the image
image = cv2.imread('parrot.png')
image = cv2.cvtColor(image, cv2.COLOR_BGR2RGB) # Convert from BGR to RGB
# Gaussian sigma
sigma=2
#parameter
amount=1.5
# Convert image to float32 for processing
image_float = image.astype(np.float32) / 255.0
# Apply Gaussian blur
blurred = cv2.GaussianBlur(image_float, (0, 0), sigmaX=sigma,
sigmaY=sigma)
# Perform Unsharp Masking
sharpened = np.clip(image_float + amount * (image_float - blurred), 0, 1)
# Plot original, blurred, and sharpened images
fig, ax = plt.subplots(1, 3, figsize=(15, 5))
ax[0].imshow(image)
ax[0].set_title("Original Image")
ax[0].axis("off")
ax[1].imshow(blurred)
ax[1].set_title("Blurred Image")
ax[1].axis("off")
ax[2].imshow(sharpened)
ax[2].set_title("Sharpened (Unsharp Masking)")
ax[2].axis("off")
plt.show()
```

```python
image = cv2.imread('monkey.png')
image = cv2.cvtColor(image, cv2.COLOR_BGR2RGB)

kernel = np.ones((7, 7), np.float32) / 49
smoothed = cv2.filter2D(image, -1, kernel)

fig, ax = plt.subplots(1, 2, figsize=(10, 5))
ax[0].imshow(image)
ax[0].set_title("Original Image")
ax[0].axis("off")
ax[1].imshow(smoothed)
ax[1].set_title("Smoothed (7x7 Box Filter)")
ax[1].axis("off")
plt.show()
```

```python
image = cv2.imread('monkey.png')
image = cv2.cvtColor(image, cv2.COLOR_BGR2RGB)

gaussian_5 = cv2.GaussianBlur(image, (5, 5), 0)
gaussian_21 = cv2.GaussianBlur(image, (21, 21), 0)

fig, ax = plt.subplots(1, 3, figsize=(15, 5))
ax[0].imshow(image)
ax[0].set_title("Original Image")
ax[0].axis("off")
ax[1].imshow(gaussian_5)
ax[1].set_title("Gaussian 5x5")
ax[1].axis("off")
ax[2].imshow(gaussian_21)
ax[2].set_title("Gaussian 21x21")
ax[2].axis("off")
plt.show()
```

```python
image = cv2.imread('monkey.png')
image = cv2.cvtColor(image, cv2.COLOR_BGR2RGB)

image_float = image.astype(np.float32) / 255.0

laplacian = cv2.Laplacian(image_float, cv2.CV_32F)

sharpened = np.clip(image_float - laplacian, 0, 1)

laplacian_display = (laplacian - laplacian.min()) / (laplacian.max() - laplacian.min())

fig, ax = plt.subplots(1, 3, figsize=(15, 5))
ax[0].imshow(image)
ax[0].set_title("Original Image")
ax[0].axis("off")
ax[1].imshow(laplacian_display)
ax[1].set_title("Laplacian Filter")
ax[1].axis("off")
ax[2].imshow(sharpened)
ax[2].set_title("Sharpened Image")
ax[2].axis("off")
plt.show()
```
